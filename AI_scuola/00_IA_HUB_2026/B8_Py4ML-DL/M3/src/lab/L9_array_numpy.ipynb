## 1. Importare NumPy

`np` è l'alias convenzionale. `np.__version__` contiene la versione installata.

```python
import numpy as np
print(np.__version__)
```

## 2. Creare array

Un array NumPy contiene valori tutti dello stesso tipo, indicato dall'attributo `dtype`.

```python
a = np.array([18.5, 21.0, 23.4, 19.8, 25.1])
print(a)
print("dtype:", a.dtype, " shape:", a.shape, " ndim:", a.ndim, " size:", a.size)

b = np.array([1, 2, 3])
print(b.dtype)          # interi
```

```python
print(np.zeros(4))                 # quattro zeri
print(np.ones((2, 3)))             # matrice 2 x 3 di uni
print(np.full(3, 7.5))             # tre valori 7.5
print(np.arange(0, 10, 2))         # come range, ma produce un array
print(np.linspace(0, 1, 5))        # 5 valori equidistanti da 0 a 1, estremi inclusi
```

## 3. Array a due dimensioni

Si creano da una lista di liste: ogni lista interna è una riga. `shape` è la tupla (righe, colonne).

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m)
print("shape:", m.shape, " ndim:", m.ndim)

v = np.arange(12)
print(v.reshape(3, 4))             # stessi 12 valori, disposti in 3 righe e 4 colonne
```

## 4. Indici e sottoarray

In due dimensioni si indicano riga e colonna separate da una virgola: `m[riga, colonna]`. I due punti `:` da soli indicano "tutte".

```python
t = np.array([[10, 11, 12, 13],
              [20, 21, 22, 23],
              [30, 31, 32, 33]])
print(t[1, 2])        # riga 1, colonna 2 -> 22
print(t[0])           # riga 0
print(t[:, 1])        # colonna 1
print(t[0:2, 1:3])    # righe 0-1, colonne 1-2
print(t[-1, -1])      # ultimo elemento
```
