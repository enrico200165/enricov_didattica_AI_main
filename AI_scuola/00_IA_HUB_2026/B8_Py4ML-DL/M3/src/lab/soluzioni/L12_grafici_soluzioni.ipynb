Eseguire le celle in ordine con `Maiusc + Invio`. Per i grafici la cella di controllo verifica i dati calcolati; il grafico va controllato guardandolo.

```python
# Funzione di controllo degli esercizi: eseguire questa cella, non modificarla.
def controlla(numero, prova):
    try:
        esito = bool(prova())
    except Exception as e:
        print(f"Esercizio {numero}: da rivedere ({type(e).__name__})")
        return
    print(f"Esercizio {numero}: " + ("corretto" if esito else "da rivedere"))
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Grafico di una funzione

Si calcolano i valori di $x$ e di $y$ come array e si passano a `plt.plot`. Le altre funzioni aggiungono titolo, etichette degli assi, griglia e legenda.

```python
x = np.linspace(-5, 5, 200)
y = 1 / (1 + np.exp(-x))

plt.plot(x, y, label="sigmoide")
plt.title("Funzione sigmoide")
plt.xlabel("z")
plt.ylabel("valore")
plt.grid(True)
plt.legend()
plt.show()
```

## 2. Più curve nello stesso grafico

```python
plt.plot(x, 1 / (1 + np.exp(-x)), label="sigmoide")
plt.plot(x, np.tanh(x), label="tanh")
plt.plot(x, np.maximum(x, 0), label="ReLU")
plt.ylim(-1.5, 3)
plt.axhline(0, color="gray", linewidth=0.8)   # asse orizzontale
plt.legend()
plt.show()
```

## 3. Punti: scatter

Due nuvole di punti generate con la distribuzione normale, attorno a due centri diversi: un piccolo insieme di dati con due classi.

```python
rng = np.random.default_rng(3)
classe_a = rng.normal(loc=[2, 2], scale=0.6, size=(30, 2))     # 30 punti attorno a (2, 2)
classe_b = rng.normal(loc=[4, 4], scale=0.6, size=(30, 2))     # 30 punti attorno a (4, 4)

plt.scatter(classe_a[:, 0], classe_a[:, 1], label="classe A")
plt.scatter(classe_b[:, 0], classe_b[:, 1], marker="s", label="classe B")
plt.xlabel("x1")
plt.ylabel("x2")
plt.legend()
plt.axis("equal")
plt.show()
```

## 4. Una retta di separazione

Un neurone con due ingressi separa il piano con la retta $w_1 x_1 + w_2 x_2 + b = 0$. Ricavando $x_2$: $x_2 = -(w_1 x_1 + b) / w_2$.

```python
w1, w2, b = 1.0, 1.0, -6.0
x1 = np.linspace(0, 6, 50)
x2 = -(w1 * x1 + b) / w2

plt.scatter(classe_a[:, 0], classe_a[:, 1], label="classe A")
plt.scatter(classe_b[:, 0], classe_b[:, 1], marker="s", label="classe B")
plt.plot(x1, x2, color="black", label="retta del neurone")
plt.xlim(0, 6)
plt.ylim(0, 6)
plt.legend()
plt.show()
```

## 5. Istogramma

`plt.hist` divide i valori in intervalli (bins) e disegna quanti valori cadono in ciascuno.

```python
campione = rng.normal(0, 1, size=1000)
plt.hist(campione, bins=30)
plt.title("1000 valori dalla distribuzione normale")
plt.show()
```

## 6. Più grafici affiancati e salvataggio

`plt.subplots(righe, colonne)` crea una figura con più riquadri (assi). `fig.savefig` salva l'immagine in un file.

```python
fig, assi = plt.subplots(1, 2, figsize=(9, 3))
assi[0].plot(x, np.maximum(x, 0))
assi[0].set_title("ReLU")
assi[1].plot(x, np.tanh(x))
assi[1].set_title("tanh")
fig.tight_layout()
fig.savefig("attivazioni.png")
plt.show()
```

**Esercizio 1 (base).** Disegnare nello stesso grafico la sigmoide e la sua derivata $\sigma(z)(1 - \sigma(z))$ per $z$ tra -6 e 6, con titolo, etichette e legenda. Assegnare i valori della derivata a `derivata`.

```python
z = np.linspace(-6, 6, 200)
sig = 1 / (1 + np.exp(-z))
derivata = sig * (1 - sig)
plt.plot(z, sig, label="sigmoide")
plt.plot(z, derivata, label="derivata")
plt.title("Sigmoide e derivata")
plt.xlabel("z")
plt.legend()
plt.grid(True)
plt.show()
```

**Esercizio 2 (base).** Simulare 10 000 lanci di due dadi, calcolare `somme`, la somma dei due dadi per ogni lancio, e disegnarne l'istogramma con `bins=np.arange(1.5, 13.5)` (un intervallo per ogni somma possibile). Quale somma è la più frequente?

**Esercizio 3 (standard).** Disegnare i quattro punti della tabella di verità (0,0), (0,1), (1,0), (1,1): i punti con uscita 1 per AND con un simbolo, gli altri con un simbolo diverso. Poi disegnare la retta del neurone AND con $w_1 = w_2 = 1$, $b = -1.5$, calcolando `x2_retta` per `x1_retta = np.linspace(-0.5, 1.5, 50)`.

```python
punti = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
uscite_and = np.array([0, 0, 0, 1])
x1_retta = np.linspace(-0.5, 1.5, 50)
x2_retta = -(1 * x1_retta - 1.5) / 1
uno = punti[uscite_and == 1]
zero = punti[uscite_and == 0]
plt.scatter(zero[:, 0], zero[:, 1], marker="o", s=80, label="uscita 0")
plt.scatter(uno[:, 0], uno[:, 1], marker="s", s=80, label="uscita 1")
plt.plot(x1_retta, x2_retta, color="black", label="retta AND")
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.legend()
plt.show()
```

```python
controlla(3, lambda: np.allclose(x2_retta, 1.5 - x1_retta))
```

**Esercizio 4 (standard).** Con le nuvole `classe_a` e `classe_b` della sezione 3 calcolare la frazione di punti classificati correttamente dal neurone con $w_1 = w_2 = 1$, $b = -6$: un punto è assegnato alla classe B se $w_1 x_1 + w_2 x_2 + b \geq 0$. Assegnare il risultato ad `accuratezza` (valore tra 0 e 1). Suggerimento: `classe_b @ np.array([w1, w2]) + b >= 0` è la maschera dei punti B classificati bene.

```python
w = np.array([1.0, 1.0])
corretti_b = (classe_b @ w - 6 >= 0).sum()
corretti_a = (classe_a @ w - 6 < 0).sum()
accuratezza = (corretti_a + corretti_b) / (len(classe_a) + len(classe_b))
print(accuratezza)
```

```python
_w = np.array([1.0, 1.0])
_atteso = (((classe_b @ _w - 6) >= 0).sum() + ((classe_a @ _w - 6) < 0).sum()) / 60
controlla(4, lambda: np.isclose(accuratezza, _atteso))
```

**Esercizio 5 (approfondimento).** Regione di decisione. `np.meshgrid` crea una griglia di punti del piano: `X1, X2 = np.meshgrid(a, b)` restituisce due matrici con le coordinate di tutti i punti della griglia. Calcolare l'uscita della sigmoide del neurone ($w_1 = w_2 = 1$, $b = -6$) su tutta la griglia e disegnarla con `plt.contourf(X1, X2, uscita, levels=20)`, sovrapponendo le due nuvole di punti. Assegnare le uscite a `uscita`.

```python
g = np.linspace(0, 6, 100)
X1, X2 = np.meshgrid(g, g)
uscita = 1 / (1 + np.exp(-(X1 + X2 - 6)))
plt.contourf(X1, X2, uscita, levels=20, cmap="Greys")
plt.colorbar(label="uscita del neurone")
plt.scatter(classe_a[:, 0], classe_a[:, 1], label="classe A")
plt.scatter(classe_b[:, 0], classe_b[:, 1], marker="s", label="classe B")
plt.legend()
plt.show()
```
