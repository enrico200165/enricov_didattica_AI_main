## 1. Prodotto scalare

Il prodotto scalare di due vettori della stessa lunghezza è la somma dei prodotti degli elementi corrispondenti: è la somma pesata della lezione L5. In NumPy: `np.dot(a, b)` oppure `a @ b`.

```python
x = np.array([1.0, 0.0, 1.0])       # ingressi
w = np.array([0.6, 0.6, 0.6])       # pesi
print(np.dot(x, w))
print(x @ w)
print((x * w).sum())                 # stessa cosa: prodotti elemento per elemento, poi somma
```

## 2. Norma e angolo

La norma (lunghezza) di un vettore è $\sqrt{v_1^2 + v_2^2 + \dots}$. Il prodotto scalare è positivo se i vettori formano un angolo acuto, zero se sono perpendicolari, negativo se l'angolo è ottuso.

```python
a = np.array([3.0, 4.0])
print(np.linalg.norm(a))                       # 5.0
print(np.array([1, 0]) @ np.array([1, 1]))     # angolo di 45 gradi: positivo
print(np.array([1, 0]) @ np.array([0, 1]))     # perpendicolari: zero
print(np.array([1, 0]) @ np.array([-1, 1]))    # angolo di 135 gradi: negativo
```

## 3. Matrici, forma, trasposta

```python
W = np.array([[0.2, -0.5, 1.0],
              [0.7,  0.1, -0.3]])
print(W.shape)       # (2, 3)
print(W.T)           # trasposta: righe e colonne scambiate
print(W.T.shape)     # (3, 2)
```

## 4. Prodotto matrice-vettore

Ogni elemento del risultato è il prodotto scalare di una riga della matrice con il vettore. La matrice $2 \times 3$ per un vettore di 3 elementi dà un vettore di 2 elementi.

```python
x = np.array([1.0, 2.0, 3.0])
print(W @ x)
print(W[0] @ x, W[1] @ x)            # le due righe, una alla volta: stesso risultato
```

## 5. Forme compatibili

In `A @ B` il numero di colonne di `A` deve essere uguale al numero di righe di `B`. Altrimenti NumPy segnala un errore.

```python
try:
    W @ np.array([1.0, 2.0])
except ValueError as e:
    print("ValueError:", e)
```

## 6. Uno strato di neuroni in una riga

Con la matrice dei pesi `W` (una riga per neurone) e il vettore dei bias `b`, le uscite di tutti i neuroni dello strato sono `attivazione(W @ x + b)`.

```python
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

b = np.array([0.1, -0.2])
z = W @ x + b
print("somme pesate:", z)
print("uscite:", sigmoide(z))
```

**Esercizio 3 (standard).** Neurone con attivazione a gradino, in forma vettoriale: completare `neurone(x, w, b)` che restituisce 1 se `w @ x + b >= 0`, altrimenti 0. Verificare che con `w = [1, 1]` e `b = -1.5` si ottenga AND.

```python
def neurone(x, w, b):
    return 1 if w @ x + b >= 0 else 0
```

```python
w_and = np.array([1.0, 1.0])
controlla(3, lambda: [neurone(np.array([a, c]), w_and, -1.5) for a in [0, 1] for c in [0, 1]] == [0, 0, 0, 1])
```

**Esercizio 4 (standard).** Creare lo strato con 3 ingressi e 4 neuroni: matrice dei pesi `W4` di forma (4, 3) con valori casuali normali (media 0, deviazione standard 1, seme 1) e bias `b4` di 4 zeri. Calcolare `uscite4`, le uscite con attivazione ReLU per l'ingresso `x4`.

```python
rng = np.random.default_rng(1)
x4 = np.array([0.5, -1.0, 2.0])
W4 = rng.normal(0, 1, size=(4, 3))
b4 = np.zeros(4)
uscite4 = np.maximum(W4 @ x4 + b4, 0)
print(uscite4)
```

**Esercizio 6 (approfondimento).** La rete XOR della lezione L8 in forma matriciale. Strato nascosto: neuroni OR e NAND, cioè pesi `[[1, 1], [-1, -1]]` e bias `[-0.5, 1.5]`. Uscita: neurone AND sui due valori nascosti. Completare `xor(x)` usando `@` e `np.where` per il gradino.

```python
W_h = np.array([[1, 1], [-1, -1]])
b_h = np.array([-0.5, 1.5])
w_o = np.array([1, 1])
b_o = -1.5

def xor(x):
    h = np.where(W_h @ x + b_h >= 0, 1, 0)
    return 1 if w_o @ h + b_o >= 0 else 0
```

```python
controlla(6, lambda: [xor(np.array([a, c])) for a in [0, 1] for c in [0, 1]] == [0, 1, 1, 0])
```
