Un notebook è composto da **celle**. Ci sono due tipi di cella:

- celle di **codice**: contengono istruzioni Python; si eseguono con `Maiusc + Invio` (in inglese `Shift + Enter`)
- celle di **testo**: contengono testo scritto in Markdown, come questa

Eseguire le celle una alla volta, dall'alto in basso.

```python
print("Ciao dal notebook")
```

## 1. Il risultato di una cella

Se l'ultima riga di una cella di codice è un'espressione, il notebook ne mostra il valore sotto la cella, anche senza `print()`. Le espressioni che precedono l'ultima riga vengono calcolate ma non mostrate.

```python
2 + 3
10 * 4
```

## 2. Lo stato del kernel

Il **kernel** è il programma Python che esegue le celle. Le variabili create in una cella restano in memoria e si possono usare nelle celle successive.

```python
raggio = 5
```

```python
area = 3.14159 * raggio ** 2
area
```
