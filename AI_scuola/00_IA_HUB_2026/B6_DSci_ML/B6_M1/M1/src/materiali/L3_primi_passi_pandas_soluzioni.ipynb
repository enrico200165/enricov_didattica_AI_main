## 0. Preparazione

La cella seguente importa la libreria pandas con il nome breve `pd` e definisce la funzione `controlla`, che stampa "Corretto." quando un esercizio è risolto.

```python
import pandas as pd

def controlla(condizione, suggerimento="Non ancora: rileggere la consegna."):
    """Stampa un riscontro immediato sull'esercizio."""
    print("Corretto." if condizione else suggerimento)
```

## 3. Condizioni numeriche (esercizio 3)

Operatori di confronto: `==` uguale, `!=` diverso, `>` maggiore, `>=` maggiore o uguale, `<` minore, `<=` minore o uguale.

```python
pesanti = pinguini[pinguini["massa_g"] > 5000]
print("pinguini con massa superiore a 5000 g:", len(pesanti))
pesanti["specie"].value_counts()
```
